### 1. First-time Crop Recommendation Model (Random Forest / XGBoost)

This section will train a classification model to recommend an initial crop based on the soil analysis (N, P, K, pH, Soil Type) and environmental factors (Rainfall, Temperature, Humidity). We'll use a Random Forest Classifier for this task.

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder

# Prepare the data
X = df_crop_rec.drop('Crop_Recommended', axis=1)
y = df_crop_rec['Crop_Recommended']

# Encode categorical features (Soil_Type) and target (Crop_Recommended)
# The new Kaggle dataset does not have a 'Soil_Type' column, so this line is removed.
# X = pd.get_dummies(X, columns=['Soil_Type'], drop_first=True)

# Use LabelEncoder for the target variable
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Number of unique crops: {len(le.classes_)}")
```

```python
# Initialize and train the Random Forest Classifier with updated hyperparameters
rf_model = RandomForestClassifier(n_estimators=200, max_depth=15, min_samples_split=5, min_samples_leaf=2, class_weight='balanced', random_state=42)
rf_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = rf_model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=le.classes_)

print(f"Random Forest Classifier Accuracy (with updated hyperparameters): {accuracy:.4f}")
print("\nClassification Report (with updated hyperparameters):")
print(report)
```

### 2. Soil Nutrient Transition Model

This is a crucial component for the reinforcement learning part. Since we don't have real-world data for how specific crops affect soil nutrients, we will create a *simplified, conceptual model*.

This model will estimate the change in Nitrogen (N), Phosphorus (P), and Potassium (K) after a specific crop is grown. We'll make some assumptions:

*   Each crop consumes a certain amount of N, P, K from the soil.
*   There's a natural replenishment/degradation over time (a small baseline change).

**Simplifications and Assumptions:**

*   **Nutrient Consumption:** We'll define a dictionary `CROP_NUTRIENT_CONSUMPTION` that estimates how much N, P, and K a crop 'removes' from the soil. These values are illustrative.
*   **pH and other factors:** For simplicity in this conceptual model, we'll assume pH, Rainfall, Temperature, and Humidity remain constant or have minor/random fluctuations not directly modeled by crop growth. In a real scenario, these would also evolve.
*   **Yield/Efficiency:** We are not modeling yield directly here, just nutrient changes. The RL reward function will implicitly link to 'good' soil states.



```python
# Define conceptual nutrient consumption per crop (illustrative values for the new Kaggle dataset crops)
# These values are illustrative and should be replaced with actual agricultural data for real-world applications.
CROP_NUTRIENT_CONSUMPTION = {
    'rice': {'N': 40, 'P': 15, 'K': 30},
    'maize': {'N': 50, 'P': 20, 'K': 40},
    'chickpea': {'N': -15, 'P': 10, 'K': 20}, # Nitrogen fixer
    'kidneybeans': {'N': -10, 'P': 10, 'K': 18}, # Nitrogen fixer
    'pigeonpeas': {'N': -12, 'P': 10, 'K': 22}, # Nitrogen fixer
    'mothbeans': {'N': -8, 'P': 8, 'K': 15}, # Nitrogen fixer
    'mungbean': {'N': -10, 'P': 9, 'K': 17}, # Nitrogen fixer
    'blackgram': {'N': -12, 'P': 10, 'K': 19}, # Nitrogen fixer
    'lentil': {'N': -10, 'P': 9, 'K': 16}, # Nitrogen fixer
    'pomegranate': {'N': 30, 'P': 12, 'K': 25},
    'banana': {'N': 60, 'P': 25, 'K': 50},
    'mango': {'N': 45, 'P': 20, 'K': 35},
    'grapes': {'N': 35, 'P': 15, 'K': 28},
    'watermelon': {'N': 40, 'P': 18, 'K': 32},
    'muskmelon': {'N': 38, 'P': 16, 'K': 30},
    'apple': {'N': 35, 'P': 15, 'K': 25},
    'orange': {'N': 50, 'P': 20, 'K': 40},
    'papaya': {'N': 40, 'P': 18, 'K': 30},
    'coconut': {'N': 25, 'P': 10, 'K': 20},
    'cotton': {'N': 60, 'P': 25, 'K': 45},
    'jute': {'N': 55, 'P': 22, 'K': 40},
    'coffee': {'N': 70, 'P': 30, 'K': 55}
}

# Define the function to update soil nutrients
def update_soil_nutrients(current_soil_state, crop_planted_name):
    # current_soil_state is a pandas Series or dict with N, P, K, pH

    # Make a copy to avoid modifying the original state
    next_soil_state = current_soil_state.copy()

    if crop_planted_name in CROP_NUTRIENT_CONSUMPTION:
        consumption = CROP_NUTRIENT_CONSUMPTION[crop_planted_name]
        next_soil_state['Nitrogen_soil'] -= consumption.get('N', 0)
        next_soil_state['Phosphorus_soil'] -= consumption.get('P', 0)
        next_soil_state['Potassium_soil'] -= consumption.get('K', 0)
    else:
        # If crop not in dictionary, assume some default consumption or no change
        pass

    # Add a small natural regeneration/degradation factor (illustrative)
    next_soil_state['Nitrogen_soil'] = max(0, next_soil_state['Nitrogen_soil'] + np.random.uniform(-5, 10))
    next_soil_state['Phosphorus_soil'] = max(0, next_soil_state['Phosphorus_soil'] + np.random.uniform(-2, 5))
    next_soil_state['Potassium_soil'] = max(0, next_soil_state['Potassium_soil'] + np.random.uniform(-5, 10))

    # Ensure nutrients don't go below zero (or some reasonable minimum)
    next_soil_state['Nitrogen_soil'] = np.clip(next_soil_state['Nitrogen_soil'], 0, 200) # Capping for realism
    next_soil_state['Phosphorus_soil'] = np.clip(next_soil_state['Phosphorus_soil'], 0, 150)
    next_soil_state['Potassium_soil'] = np.clip(next_soil_state['Potassium_soil'], 0, 300)

    # pH, Rainfall, Temperature, Humidity remain largely unchanged for simplicity in this model
    # In a more advanced model, pH could also shift, and weather would vary seasonally

    return next_soil_state

print("Soil nutrient transition function `update_soil_nutrients` defined.")
```

### 3. Reinforcement Learning Environment

Now, we'll define a custom OpenAI Gym-like environment. This environment will simulate the farm, where the agent chooses a crop (action) and receives a new soil state and a reward.

**Environment Components:**

*   **Observation Space:** The current soil and environmental conditions (N, P, K, pH, Rainfall, Temperature, Humidity).
*   **Action Space:** The set of all possible crops the farmer can plant.
*   **Reward Function:** This is critical. For this conceptual project, we will define a reward based on:
    *   **Positive Reward:** If the chosen crop is considered 'suitable' for the current soil conditions. We can derive suitability from the `rf_model` (predicting if the RF model would recommend this crop).
    *   **Negative Reward (Penalty):** If the chosen crop is 'unsuitable' or leads to very low nutrient levels.
    *   **Nutrient Balance Reward:** A small positive reward for keeping nutrients within a healthy range, or a penalty for extreme depletion.
*   **Done Condition:** The episode ends after a fixed number of 'seasons' or crop cycles.



```python
import gym
from gym import spaces
import random
import numpy as np
import pandas as pd

class CropRecommendationEnv(gym.Env):
    def __init__(self, rf_model, label_encoder, initial_soil_data, crops_list):
        super(CropRecommendationEnv, self).__init__()
        self.rf_model = rf_model
        self.label_encoder = label_encoder
        self.crops_list = crops_list
        self.initial_soil_data = initial_soil_data # Use this for reset to original distribution

        # Get feature names for RF model input (no Soil_Type in new dataset)
        self.rf_features = rf_model.feature_names_in_ if hasattr(rf_model, 'feature_names_in_') else initial_soil_data.drop(columns=['Crop_Recommended']).columns

        # Define observation space dynamically based on the number of RF features
        # Numerical features: Nitrogen_soil, Phosphorus_soil, Potassium_soil, pH_soil, Rainfall_mm, Temperature_C, Humidity_%

        # Define low and high bounds for each feature in the exact order of rf_features
        # Based on typical ranges for the Kaggle dataset's features
        low_bounds = [
            0,   # Nitrogen_soil
            0,   # Phosphorus_soil
            0,   # Potassium_soil
            3.0, # pH_soil
            20,  # Rainfall_mm
            0,   # Temperature_C
            10   # Humidity_%
        ]
        high_bounds = [
            150, # Nitrogen_soil
            150, # Phosphorus_soil
            250, # Potassium_soil
            10.0,# pH_soil
            300, # Rainfall_mm
            45,  # Temperature_C
            100  # Humidity_%
        ]

        self.observation_space = spaces.Box(
            low=np.array(low_bounds, dtype=np.float32),
            high=np.array(high_bounds, dtype=np.float32),
            shape=(len(self.rf_features),),
            dtype=np.float32
        )

        # Define action space (number of available crops)
        self.action_space = spaces.Discrete(len(self.crops_list))

        self.current_soil_state = None
        self.current_step = 0
        self.max_steps = 20 # Increased to allow more time for optimization

        # Define optimal nutrient ranges for dense rewards
        self.optimal_n_range = (50, 120)
        self.optimal_p_range = (20, 80)
        self.optimal_k_range = (80, 200)
        self.optimal_ph_range = (6.0, 7.5)

    def _get_obs(self):
        # Ensure the order matches the RF model's expected input features
        obs_dict = {
            'Nitrogen_soil': self.current_soil_state['Nitrogen_soil'],
            'Phosphorus_soil': self.current_soil_state['Phosphorus_soil'],
            'Potassium_soil': self.current_soil_state['Potassium_soil'],
            'pH_soil': self.current_soil_state['pH_soil'],
            'Rainfall_mm': self.current_soil_state['Rainfall_mm'],
            'Temperature_C': self.current_soil_state['Temperature_C'],
            'Humidity_%': self.current_soil_state['Humidity_%'],
            # 'Soil_Type': self.current_soil_state['Soil_Type'] # No Soil_Type in new dataset
        }

        # Create a temporary DataFrame to ensure proper feature order for the RF model.
        temp_df = pd.DataFrame([obs_dict])

        # Reindex to ensure all expected columns (and their order) from self.rf_features are present
        final_obs_df = temp_df.reindex(columns=self.rf_features, fill_value=0.0)

        return final_obs_df.iloc[0].values.astype(np.float32)

    def _calculate_distance_to_optimal(self, value, optimal_range):
        min_val, max_val = optimal_range
        if value < min_val:
            return min_val - value
        elif value > max_val:
            return value - max_val
        else:
            return 0 # Inside optimal range

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        # Randomly pick an initial soil state from the dataset
        random_idx = self.np_random.integers(0, len(self.initial_soil_data))
        self.current_soil_state = self.initial_soil_data.iloc[random_idx].copy()

        self.current_step = 0
        observation = self._get_obs()
        info = {}
        return observation, info

    def step(self, action):
        self.current_step += 1

        crop_chosen_name = self.crops_list[action]

        # Store previous state for dense reward calculation
        previous_soil_state = self.current_soil_state.copy()

        # Update soil nutrients based on the chosen crop
        self.current_soil_state = update_soil_nutrients(self.current_soil_state, crop_chosen_name)

        # Calculate reward
        reward = 0

        # Predict what the RF model would recommend for the *current* soil state
        # Need to prepare input for RF model exactly as X was prepared
        rf_prediction_input_dict = {
            'Nitrogen_soil': self.current_soil_state['Nitrogen_soil'],
            'Phosphorus_soil': self.current_soil_state['Phosphorus_soil'],
            'Potassium_soil': self.current_soil_state['Potassium_soil'],
            'pH_soil': self.current_soil_state['pH_soil'],
            'Rainfall_mm': self.current_soil_state['Rainfall_mm'],
            'Temperature_C': self.current_soil_state['Temperature_C'],
            'Humidity_%': self.current_soil_state['Humidity_%'],
            # 'Soil_Type': self.current_soil_state['Soil_Type'] # No Soil_Type in new dataset
        }

        temp_df = pd.DataFrame([rf_prediction_input_dict])
        # No Soil_Type to one-hot encode
        rf_prediction_input = temp_df.reindex(columns=self.rf_features, fill_value=0.0)

        rf_recommended_crop_encoded = self.rf_model.predict(rf_prediction_input)[0]
        rf_recommended_crop_name = self.label_encoder.inverse_transform([rf_recommended_crop_encoded])[0]

        # Reward 1: If the chosen crop matches the RF model's recommendation (reduced impact)
        if crop_chosen_name == rf_recommended_crop_name:
            reward += 5
        else:
            reward -= 1

        # Determine if current state is 'successful' (all optimal nutrients)
        n_ok = self.optimal_n_range[0] <= self.current_soil_state['Nitrogen_soil'] <= self.optimal_n_range[1]
        p_ok = self.optimal_p_range[0] <= self.current_soil_state['Phosphorus_soil'] <= self.optimal_p_range[1]
        k_ok = self.optimal_k_range[0] <= self.current_soil_state['Potassium_soil'] <= self.optimal_k_range[1]
        ph_ok = self.optimal_ph_range[0] <= self.current_soil_state['pH_soil'] <= self.optimal_ph_range[1]

        # Reward 2: Nutrient level check (incentivize balanced soil significantly)
        if n_ok and p_ok and k_ok and ph_ok:
            reward += 50 # Strong positive reward for optimal soil health
        else:
            reward -= 5 # Increased penalty for general sub-optimal state

        # Dense Reward: Encourage moving closer to optimal ranges for N, P, K, pH
        nutrient_keys = ['Nitrogen_soil', 'Phosphorus_soil', 'Potassium_soil', 'pH_soil']
        optimal_ranges = {
            'Nitrogen_soil': self.optimal_n_range,
            'Phosphorus_soil': self.optimal_p_range,
            'Potassium_soil': self.optimal_k_range,
            'pH_soil': self.optimal_ph_range
        }

        for key in nutrient_keys:
            prev_dist = self._calculate_distance_to_optimal(previous_soil_state[key], optimal_ranges[key])
            current_dist = self._calculate_distance_to_optimal(self.current_soil_state[key], optimal_ranges[key])

            if current_dist < prev_dist: # Moving closer
                reward += 0.5
            elif current_dist > prev_dist: # Moving further away
                reward -= 0.5

        # Bonus for planting nitrogen-fixing crops when Nitrogen is low
        # CROP_NUTRIENT_CONSUMPTION is a global variable from ad510374
        global CROP_NUTRIENT_CONSUMPTION
        nitrogen_consumption = CROP_NUTRIENT_CONSUMPTION.get(crop_chosen_name, {}).get('N', 0)
        if nitrogen_consumption < 0 and previous_soil_state['Nitrogen_soil'] < self.optimal_n_range[0]:
            reward += 15

        # Penalize severely low nutrients (increased penalty)
        if self.current_soil_state['Nitrogen_soil'] < 10 or \
           self.current_soil_state['Phosphorus_soil'] < 5 or \
           self.current_soil_state['Potassium_soil'] < 20:
            reward -= 10

        # Very high penalty for critical nutrient depletion (hitting zero, increased penalty)
        if self.current_soil_state['Nitrogen_soil'] <= 0 or \
           self.current_soil_state['Phosphorus_soil'] <= 0 or \
           self.current_soil_state['Potassium_soil'] <= 0:
            reward -= 25

        # Check if episode is done
        done = self.current_step >= self.max_steps

        episode_successful_flag = n_ok and p_ok and k_ok and ph_ok and (reward > 0 or self.current_step == self.max_steps) # Define success for terminal reward

        # Terminal Reward for overall success (if episode ends well)
        if done and episode_successful_flag:
            reward += 200 # Large bonus for successfully completing an episode with optimal conditions

        observation = self._get_obs()
        info = {
            'crop_chosen': crop_chosen_name,
            'rf_recommended': rf_recommended_crop_name,
            'soil_N': self.current_soil_state['Nitrogen_soil'],
            'soil_P': self.current_soil_state['Phosphorus_soil'],
            'soil_K': self.current_soil_state['Potassium_soil'],
            'soil_pH': self.current_soil_state['pH_soil']
        }

        # Handle truncation (optional, for gym environments with max_steps)
        truncated = done # In this case, truncation is based on max_steps

        return observation, reward, done, truncated, info

    def render(self):
        pass # Not implementing rendering for this text-based environment

    def close(self):
        pass

print("Custom Gym Environment `CropRecommendationEnv` defined.")
```

```python
import sys
import pandas as pd
import numpy as np
import kagglehub
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
import gym
from gym import spaces
import random

from stable_baselines3 import DQN
from stable_baselines3.common.env_util import make_vec_env

# --- Ensure all dependencies are initialized --- BEGIN

# 1. Ensure df_crop_rec and crops list are defined (from cell e2bfeedf)
if 'df_crop_rec' not in globals() or 'crops' not in globals():
    print("Loading Kaggle dataset and defining 'crops' list...")
    try:
        path = kagglehub.dataset_download("atharvaingle/crop-recommendation-dataset")
        csv_path = f"{path}/Crop_recommendation.csv"
        df_crop_rec = pd.read_csv(csv_path)
        df_crop_rec = df_crop_rec.rename(columns={
            'N': 'Nitrogen_soil', 'P': 'Phosphorus_soil', 'K': 'Potassium_soil',
            'temperature': 'Temperature_C', 'humidity': 'Humidity_%', 'ph': 'pH_soil',
            'rainfall': 'Rainfall_mm', 'label': 'Crop_Recommended'
        })
        if 'Crop_Recommended' in df_crop_rec.columns:
            crops = df_crop_rec['Crop_Recommended'].unique().tolist()
            print(f"'crops' list defined: {crops}")
        else:
            raise ValueError("Could not find 'Crop_Recommended' column in df_crop_rec after loading.")
    except Exception as e:
        raise RuntimeError(f"Failed to load dataset or define 'crops' list: {e}")

# 2. Ensure LabelEncoder (le) and RandomForest model (rf_model) are trained (from cells 749cd84c, 943843ba)
if 'le' not in globals() or 'rf_model' not in globals():
    print("Training LabelEncoder and RandomForest model...")
    try:
        X = df_crop_rec.drop('Crop_Recommended', axis=1)
        y = df_crop_rec['Crop_Recommended']
        le = LabelEncoder()
        y_encoded = le.fit_transform(y)
        X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
        rf_model = RandomForestClassifier(n_estimators=200, max_depth=15, min_samples_split=5, min_samples_leaf=2, class_weight='balanced', random_state=42)
        rf_model.fit(X_train, y_train)
        print("LabelEncoder and RandomForest model trained.")
    except Exception as e:
        raise RuntimeError(f"Failed to train RF model or LabelEncoder: {e}")

# 3. Ensure CROP_NUTRIENT_CONSUMPTION and update_soil_nutrients function are defined (from cell ad510374)
if 'CROP_NUTRIENT_CONSUMPTION' not in globals() or 'update_soil_nutrients' not in globals():
    print("Defining CROP_NUTRIENT_CONSUMPTION and update_soil_nutrients function...")
    CROP_NUTRIENT_CONSUMPTION = {
        'rice': {'N': 40, 'P': 15, 'K': 30},
        'maize': {'N': 50, 'P': 20, 'K': 40},
        'chickpea': {'N': -15, 'P': 10, 'K': 20},
        'kidneybeans': {'N': -10, 'P': 10, 'K': 18},
        'pigeonpeas': {'N': -12, 'P': 10, 'K': 22},
        'mothbeans': {'N': -8, 'P': 8, 'K': 15},
        'mungbean': {'N': -10, 'P': 9, 'K': 17},
        'blackgram': {'N': -12, 'P': 10, 'K': 19},
        'lentil': {'N': -10, 'P': 9, 'K': 16},
        'pomegranate': {'N': 30, 'P': 12, 'K': 25},
        'banana': {'N': 60, 'P': 25, 'K': 50},
        'mango': {'N': 45, 'P': 20, 'K': 35},
        'grapes': {'N': 35, 'P': 15, 'K': 28},
        'watermelon': {'N': 40, 'P': 18, 'K': 32},
        'muskmelon': {'N': 38, 'P': 16, 'K': 30},
        'apple': {'N': 35, 'P': 15, 'K': 25},
        'orange': {'N': 50, 'P': 20, 'K': 40},
        'papaya': {'N': 40, 'P': 18, 'K': 30},
        'coconut': {'N': 25, 'P': 10, 'K': 20},
        'cotton': {'N': 60, 'P': 25, 'K': 45},
        'jute': {'N': 55, 'P': 22, 'K': 40},
        'coffee': {'N': 70, 'P': 30, 'K': 55}
    }

    def update_soil_nutrients(current_soil_state, crop_planted_name):
        next_soil_state = current_soil_state.copy()
        if crop_planted_name in CROP_NUTRIENT_CONSUMPTION:
            consumption = CROP_NUTRIENT_CONSUMPTION[crop_planted_name]
            next_soil_state['Nitrogen_soil'] -= consumption.get('N', 0)
            next_soil_state['Phosphorus_soil'] -= consumption.get('P', 0)
            next_soil_state['Potassium_soil'] -= consumption.get('K', 0)
        next_soil_state['Nitrogen_soil'] = max(0, next_soil_state['Nitrogen_soil'] + np.random.uniform(-5, 10))
        next_soil_state['Phosphorus_soil'] = max(0, next_soil_state['Phosphorus_soil'] + np.random.uniform(-2, 5))
        next_soil_state['Potassium_soil'] = max(0, next_soil_state['Potassium_soil'] + np.random.uniform(-5, 10))
        next_soil_state['Nitrogen_soil'] = np.clip(next_soil_state['Nitrogen_soil'], 0, 200)
        next_soil_state['Phosphorus_soil'] = np.clip(next_soil_state['Phosphorus_soil'], 0, 150)
        next_soil_state['Potassium_soil'] = np.clip(next_soil_state['Potassium_soil'], 0, 300)
        return next_soil_state
    print("CROP_NUTRIENT_CONSUMPTION and update_soil_nutrients function defined.")

# 4. Ensure CropRecommendationEnv class is defined (from cell ac190899)
if 'CropRecommendationEnv' not in globals():
    print("Defining CropRecommendationEnv class...")
    class CropRecommendationEnv(gym.Env):
        def __init__(self, rf_model, label_encoder, initial_soil_data, crops_list):
            super(CropRecommendationEnv, self).__init__()
            self.rf_model = rf_model
            self.label_encoder = label_encoder
            self.crops_list = crops_list
            self.initial_soil_data = initial_soil_data
            self.rf_features = rf_model.feature_names_in_ if hasattr(rf_model, 'feature_names_in_') else initial_soil_data.drop(columns=['Crop_Recommended']).columns
            low_bounds = [0, 0, 0, 3.0, 20, 0, 10]
            high_bounds = [150, 150, 250, 10.0, 300, 45, 100]
            self.observation_space = spaces.Box(
                low=np.array(low_bounds, dtype=np.float32),
                high=np.array(high_bounds, dtype=np.float32),
                shape=(len(self.rf_features),),
                dtype=np.float32
            )
            self.action_space = spaces.Discrete(len(self.crops_list))
            self.current_soil_state = None
            self.current_step = 0
            self.max_steps = 10
            self.optimal_n_range = (50, 120)
            self.optimal_p_range = (20, 80)
            self.optimal_k_range = (80, 200)
            self.optimal_ph_range = (6.0, 7.5)

        def _get_obs(self):
            obs_dict = {
                'Nitrogen_soil': self.current_soil_state['Nitrogen_soil'],
                'Phosphorus_soil': self.current_soil_state['Phosphorus_soil'],
                'Potassium_soil': self.current_soil_state['Potassium_soil'],
                'pH_soil': self.current_soil_state['pH_soil'],
                'Rainfall_mm': self.current_soil_state['Rainfall_mm'],
                'Temperature_C': self.current_soil_state['Temperature_C'],
                'Humidity_%': self.current_soil_state['Humidity_%'],
            }
            temp_df = pd.DataFrame([obs_dict])
            final_obs_df = temp_df.reindex(columns=self.rf_features, fill_value=0.0)
            return final_obs_df.iloc[0].values.astype(np.float32)

        def _calculate_distance_to_optimal(self, value, optimal_range):
            min_val, max_val = optimal_range
            if value < min_val: return min_val - value
            elif value > max_val: return value - max_val
            else: return 0

        def reset(self, seed=None, options=None):
            super().reset(seed=seed)
            random_idx = self.np_random.integers(0, len(self.initial_soil_data))
            self.current_soil_state = self.initial_soil_data.iloc[random_idx].copy()
            self.current_step = 0
            observation = self._get_obs()
            info = {}
            return observation, info

        def step(self, action):
            self.current_step += 1
            crop_chosen_name = self.crops_list[action]
            previous_soil_state = self.current_soil_state.copy()
            self.current_soil_state = update_soil_nutrients(self.current_soil_state, crop_chosen_name)
            reward = 0
            rf_prediction_input_dict = {
                'Nitrogen_soil': self.current_soil_state['Nitrogen_soil'], 'Phosphorus_soil': self.current_soil_state['Phosphorus_soil'],
                'Potassium_soil': self.current_soil_state['Potassium_soil'], 'pH_soil': self.current_soil_state['pH_soil'],
                'Rainfall_mm': self.current_soil_state['Rainfall_mm'], 'Temperature_C': self.current_soil_state['Temperature_C'],
                'Humidity_%': self.current_soil_state['Humidity_%'],
            }
            temp_df = pd.DataFrame([rf_prediction_input_dict])
            rf_prediction_input = temp_df.reindex(columns=self.rf_features, fill_value=0.0)
            rf_recommended_crop_encoded = self.rf_model.predict(rf_prediction_input)[0]
            rf_recommended_crop_name = self.label_encoder.inverse_transform([rf_recommended_crop_encoded])[0]

            if crop_chosen_name == rf_recommended_crop_name: reward += 10
            else: reward -= 1

            n_ok = self.optimal_n_range[0] <= self.current_soil_state['Nitrogen_soil'] <= self.optimal_n_range[1]
            p_ok = self.optimal_p_range[0] <= self.current_soil_state['Phosphorus_soil'] <= self.optimal_p_range[1]
            k_ok = self.optimal_k_range[0] <= self.current_soil_state['Potassium_soil'] <= self.optimal_k_range[1]
            ph_ok = self.optimal_ph_range[0] <= self.current_soil_state['pH_soil'] <= self.optimal_ph_range[1]
            if n_ok and p_ok and k_ok and ph_ok: reward += 10
            else: reward -= 1

            nutrient_keys = ['Nitrogen_soil', 'Phosphorus_soil', 'Potassium_soil', 'pH_soil']
            optimal_ranges = {
                'Nitrogen_soil': self.optimal_n_range, 'Phosphorus_soil': self.optimal_p_range,
                'Potassium_soil': self.optimal_k_range, 'pH_soil': self.optimal_ph_range
            }
            for key in nutrient_keys:
                prev_dist = self._calculate_distance_to_optimal(previous_soil_state[key], optimal_ranges[key])
                current_dist = self._calculate_distance_to_optimal(self.current_soil_state[key], optimal_ranges[key])
                if current_dist < prev_dist: reward += 0.5
                elif current_dist > prev_dist: reward -= 0.5

            nitrogen_consumption = CROP_NUTRIENT_CONSUMPTION.get(crop_chosen_name, {}).get('N', 0)
            if nitrogen_consumption < 0 and previous_soil_state['Nitrogen_soil'] < self.optimal_n_range[0]: reward += 15

            if self.current_soil_state['Nitrogen_soil'] < 10 or self.current_soil_state['Phosphorus_soil'] < 5 or self.current_soil_state['Potassium_soil'] < 20: reward -= 5

            if self.current_soil_state['Nitrogen_soil'] <= 0 or self.current_soil_state['Phosphorus_soil'] <= 0 or self.current_soil_state['Potassium_soil'] <= 0: reward -= 10

            done = self.current_step >= self.max_steps
            observation = self._get_obs()
            info = {
                'crop_chosen': crop_chosen_name, 'rf_recommended': rf_recommended_crop_name,
                'soil_N': self.current_soil_state['Nitrogen_soil'], 'soil_P': self.current_soil_state['Phosphorus_soil'],
                'soil_K': self.current_soil_state['Potassium_soil'], 'soil_pH': self.current_soil_state['pH_soil']
            }
            truncated = done
            return observation, reward, done, truncated, info

        def render(self): pass
        def close(self): pass
    print("CropRecommendationEnv class defined.")

# --- Ensure all dependencies are initialized --- END

# Re-initialize the environment if it's not already defined
if 'env' not in globals() or 'model' not in globals() or 'vec_env' not in globals():
    print("Re-initializing environment and model.")

    initial_data_for_env = df_crop_rec.copy()
    if 'Crop_Recommended' in initial_data_for_env.columns:
        initial_data_for_env = initial_data_for_env.drop(columns=['Crop_Recommended'])

    env = CropRecommendationEnv(rf_model=rf_model, label_encoder=le, initial_soil_data=df_crop_rec, crops_list=crops)
    vec_env = make_vec_env(lambda: env, n_envs=1)

    model = DQN("MlpPolicy", vec_env, verbose=0,
                learning_rate=1e-4,
                buffer_size=50000,
                learning_starts=1000,
                batch_size=64,
                gamma=0.99,
                train_freq=4,
                target_update_interval=1000,
                exploration_fraction=0.1,
                exploration_initial_eps=1.0,
                exploration_final_eps=0.05,
                seed=42)

    print("DQN agent re-initialized and training with updated hyperparameters.")
    model.learn(total_timesteps=50000)
    print("DQN agent training complete.")
else:
    print("Environment and model are already initialized and trained.")
```

```python
import pandas as pd
import numpy as np
import time
from collections import Counter

# --- Ensure environment and model are initialized/trained ---
# This block attempts to re-initialize necessary components if they are not found in the current global scope.
# It assumes that `df_crop_rec`, `rf_model`, `le`, `CropRecommendationEnv`, `update_soil_nutrients`
# and the `DQN` class are defined by preceding executed cells in the notebook.

if 'env' not in globals() or 'model' not in globals():
    print("Attempting to re-initialize environment and load/train model if not found. Please ensure all setup cells (e.g., loading data, training RF, defining environment, training DQN) have been executed.")

    # Re-import necessary libraries if running this cell independently
    from stable_baselines3 import DQN
    from stable_baselines3.common.env_util import make_vec_env

    # Ensure `crops` list is available, needed for environment initialization.
    # This part tries to re-create 'crops' if not found, assuming df_crop_rec is available.
    if 'crops' not in globals() and 'df_crop_rec' in globals() and 'Crop_Recommended' in df_crop_rec.columns:
        crops = df_crop_rec['Crop_Recommended'].unique().tolist()
        print("Dynamically defined 'crops' list for metrics evaluation.")
    elif 'crops' not in globals():
        print("Warning: 'crops' list is not defined. Please ensure cell 'e2bfeedf' has been run successfully.")

    # Re-initialize the environment if `env` is not in globals or if `CropRecommendationEnv` was redefined.
    if 'CropRecommendationEnv' in globals() and 'df_crop_rec' in globals() and 'rf_model' in globals() and 'le' in globals() and 'crops' in globals():
        initial_data_for_env = df_crop_rec.copy()
        if 'Crop_Recommended' in initial_data_for_env.columns:
            initial_data_for_env = initial_data_for_env.drop(columns=['Crop_Recommended'])
        env = CropRecommendationEnv(rf_model=rf_model, label_encoder=le, initial_soil_data=df_crop_rec, crops_list=crops)
        vec_env = make_vec_env(lambda: env, n_envs=1)
        print("Environment re-initialized successfully.")

        # Re-initialize the model if `model` is not in globals.
        if 'model' not in globals():
            print("Warning: 'model' not found in globals. Initializing a new untrained DQN model for evaluation. Performance metrics may not reflect full training.")
            model = DQN("MlpPolicy", vec_env, verbose=0,
                        learning_rate=1e-4, buffer_size=10000,
                        learning_starts=1000, batch_size=32, gamma=0.99,
                        train_freq=4, target_update_interval=1000,
                        exploration_fraction=0.1, exploration_initial_eps=1.0,
                        exploration_final_eps=0.05, seed=42)
            # For a proper evaluation, a trained model should be loaded or re-trained here.
            # For demonstration, we might do a minimal training if the model is brand new.
            # model.learn(total_timesteps=100) # Uncomment for minimal training if model is truly new.
    else:
        print("Could not fully re-initialize environment and model due to missing dependencies (e.g., `rf_model`, `le`, `df_crop_rec`). Please run all preceding setup cells.")

# --- Evaluation Loop ---
num_evaluation_episodes = 50 # Increase episodes for more reliable metrics
all_episode_rewards = []
all_episode_lengths = []
success_count = 0
inference_times = []

if 'env' in globals() and 'model' in globals():
    print(f"\n--- Running RL Agent Evaluation for {num_evaluation_episodes} Episodes ---")
    for i in range(num_evaluation_episodes):
        obs, info = env.reset() # Reset returns (observation, info) now
        done = False
        episode_reward = 0
        episode_length = 0
        episode_successful = False

        while not done:
            start_time = time.time()
            action, _states = model.predict(obs, deterministic=True)
            end_time = time.time()
            inference_times.append(end_time - start_time)

            obs, reward, done, truncated, info = env.step(action) # Step returns (obs, reward, done, truncated, info)
            episode_reward += reward
            episode_length += 1

            if done or truncated:
                # Check for success condition at the end of the episode
                # This definition of success can be refined based on specific goals.
                current_N = info.get('soil_N', 0)
                current_P = info.get('soil_P', 0)
                current_K = info.get('soil_K', 0)
                current_pH = info.get('soil_pH', 0)

                # Ensure optimal ranges are defined in the environment instance
                if hasattr(env, 'optimal_n_range') and hasattr(env, 'optimal_p_range') and \
                   hasattr(env, 'optimal_k_range') and hasattr(env, 'optimal_ph_range'):

                    n_ok = env.optimal_n_range[0] <= current_N <= env.optimal_n_range[1]
                    p_ok = env.optimal_p_range[0] <= current_P <= env.optimal_p_range[1]
                    k_ok = env.optimal_k_range[0] <= current_K <= env.optimal_k_range[1]
                    ph_ok = env.optimal_ph_range[0] <= current_pH <= env.optimal_ph_range[1]

                    # Define success: all nutrients optimal AND a positive total episode reward
                    if n_ok and p_ok and k_ok and ph_ok and episode_reward > 0:
                        episode_successful = True
                        success_count += 1
                else:
                    print("Warning: Optimal nutrient ranges not found in environment. Success rate cannot be calculated robustly.")

        all_episode_rewards.append(episode_reward)
        all_episode_lengths.append(episode_length)
        # print(f"  Episode {i+1} finished. Reward: {episode_reward:.2f}, Length: {episode_length}, Success: {episode_successful}")

    # Close environments to free resources
    env.close()
    vec_env.close()

    # --- Calculate Metrics ---
    avg_cumulative_reward = np.mean(all_episode_rewards)
    reward_std_dev = np.std(all_episode_rewards)
    avg_episode_length = np.mean(all_episode_lengths)
    success_rate = (success_count / num_evaluation_episodes) * 100 if num_evaluation_episodes > 0 else 0
    avg_inference_time = np.mean(inference_times) * 1000 if inference_times else 0 # Convert to milliseconds

    print("\n--- RL Agent Performance Metrics ---")
    print(f"1. Average cumulative reward: {avg_cumulative_reward:.2f}")
    print(f"2. Success rate: {success_rate:.2f}% (defined as: episode ends with all optimal nutrients and positive total reward)")
    print(f"3. Average episode length: {avg_episode_length:.2f} steps")
    print(f"4. Reward standard deviation: {reward_std_dev:.2f}")
    print(f"5. Convergence speed: Not directly measurable from evaluation logs. This metric typically requires monitoring learning curves during training (e.g., how many steps/episodes until policy stabilizes or reward plateaus), which would require collecting and analyzing training logs.")
    print(f"6. Sample efficiency: Not directly measurable from evaluation logs. This metric refers to how much data (samples/experiences) the agent needs to learn an effective policy. It's usually assessed by comparing performance with varying amounts of training data, which is beyond the scope of this evaluation.")
    print(f"7. Average inference time: {avg_inference_time:.3f} ms per step (time taken for the agent to choose an action)")
    print(f"8. Comparison with baseline RL algorithms: Not available as no baseline algorithms were trained for comparison in this notebook.")
else:
    print("RL agent metrics could not be calculated due to missing 'env' or 'model' objects. Please ensure all previous setup and training cells have been executed successfully.")
```
